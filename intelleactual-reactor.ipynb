```python
import pandas as pd  
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("dataset.csv")
df.head()
```

```python
print(df.columns.tolist())
```

```python
df.info()
```

```python
df=df[(df['Target'] == 'Graduate')| (df['Target'] == 'Dropout')]

```

```python
df.info()
```

```python
gender_target_counts = df.groupby('Gender')['Target'].count().reset_index()
gender_target_counts.head()
```

```python

#gender_target_counts = df.groupby('Gender')['Target'].count().reset_index()

plt.figure(figsize=(10, 6))  
#sns.barplot(x='Gender', y='Target', data=gender_target_counts) 
ax=sns.barplot(x='Gender', y='Target', data=gender_target_counts) 
ax.bar_label(ax.containers[0], fontsize=10)
plt.title('Количество мужчин и женщин')  
plt.xlabel('Пол')  
plt.ylabel('Количество') 
plt.show()  
```

```python
import pandas as pd

# Фильтруем DataFrame
filtered_df = df[df['Gender'] == 0]

# Создаем кросс-таблицу
crosstab = pd.crosstab(filtered_df['Marital status'], filtered_df['Target'], normalize='index').reset_index().sort_values(by='Marital status', ascending=True)

# Выводим кросс-таблицу
print(crosstab)
```

```python
plt.figure(figsize=(10,7))

plt.bar(crosstab['Marital status'], crosstab['Dropout'])
plt.bar(crosstab['Marital status'], crosstab['Graduate'], bottom = crosstab['Dropout'])

plt.title("Семейное положение студенток")

plt.xlabel('Семейное положение')

plt.ylabel('доля студенток')

plt.legend(['Dropout', 'Graduate'])

plt.xticks(rotation = 30)

plt.show()
```

```python
import numpy as np
import math
from sklearn.model_selection import train_test_split
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
features = [['Gender'], \
            list_prizn1,\
            list_prizn2]
y=df4['Target']
metrics = pd.DataFrame(columns=['Features', 'MAE train', 'MAE test', 'MSE train', 'MSE test', 'R2 train', 'R2 test'])
coeff=pd.DataFrame(columns=['Features', 'Coefficients', 'intercept'])
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 200)

for future in features:
  x_future_test = df4.loc[:, future]
  x_train, x_test, y_train, y_test = train_test_split(x_future_test, y, test_size = 0.2, random_state=5)

#  transformer = Normalizer().fit(x_train.loc[:, x_train.columns])
  x_train = pd.DataFrame(x_train.loc[:, x_train.columns], index = x_train.index, columns = x_train.columns)
  x_test = pd.DataFrame(x_test.loc[:, x_test.columns], index = x_test.index, columns = x_test.columns)

  regr = LinearRegression(fit_intercept=True)
  regr.fit(x_train, y_train)
  pred_train = regr.predict(x_train)
  pred_test = regr.predict(x_test)
  coeff=coeff.append({'Features' :  future, 
                      'Coefficients': regr.coef_,
                      'intercept': round(regr.intercept_,5)},ignore_index=True)
  metrics = metrics.append({'Features' : future, 
                            'MAE train' : mean_absolute_error(y_train, pred_train), 
                            'MAE test' : mean_absolute_error(y_test, pred_test), 
                            'MSE train' : mean_squared_error(y_train, pred_train), 
                            'MSE test' : mean_squared_error(y_test, pred_test), 
                            'R2 train' : r2_score(y_train, pred_train), 
                            'R2 test' : r2_score(y_test, pred_test)}, ignore_index=True)

print(metrics)
print(coeff)
```
